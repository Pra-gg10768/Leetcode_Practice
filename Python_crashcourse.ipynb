Q10. Prime number

```python
num = int(input("Enter the number: "))

if num <= 1:
    print("Not prime")
else:
    is_prime = True

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print("Prime")
    else:
        print("Not prime")
```

Q11. Find the largest number

```python
numbers = [12, 45, 7, 89, 23, 56]
largest = numbers[0]
for i in numbers:
  if(i>largest):
    largest = i
print(largest)
```

Q12. Find the second largest

```python
numbers = [12, 45, 7, 89, 23, 56]
largest = numbers[0]
second_largest = numbers[0]
for i in numbers:
  if(i>largest):
    second_largest = largest
    largest = i
print(second_largest)

```

Bonus 2 — Count even and odd digits

```python
num = int(input("Enter the number: "))
e_count = 0
o_count = 0
while(num!=0):
  b = num%10
  if(b%2==0):
    e_count +=1
  else:
    o_count += 1
  num = num//10
print("Even Digits: ",e_count)
print("Odd Digits: ",o_count)
```

Bonus 3 — Second largest distinct

```python
numbers = [10, 10, 5]
largest_distinct = float('-inf')
second_largest_distinct = float('-inf')
for i in numbers:
    if i > largest_distinct:
        second_largest_distinct = largest_distinct
        largest_distinct = i
    elif i > second_largest_distinct and i != largest_distinct:
        second_largest_distinct = i

print(second_largest_distinct)
```

```python
class Solution(object):
    def isPalindrome(self, x):
        temp = x
        a = 0
        while(temp!=0):
            b = temp%10
            a = a*10 + b
            temp = temp//10
        if (x == a):
            return True
        else:
            return False
```
